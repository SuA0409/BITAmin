```python
import os
import json
import pandas as pd
import spacy
from tqdm import tqdm
```

```python
# loading

nlp = spacy.load("en_core_sci_sm", disable=["tagger", "lemmatizer"])  # scispaCy로 문장/토큰 분리
data = pd.read_csv("C:/Users/user/Desktop/BITAmin 15/4_25-2/NLP project/merged_dataset.csv")
```

```python
# title + abstract 합쳐서 리스트 생성
texts = [(str(row['title'] or "") + " " + str(row['abstract'] or "")).strip()
         for _, row in data.reset_index(drop=True).iterrows()]
doc_keys = [f"doc_{i}" for i in range(len(texts))]

# nlp.pipe()로 병렬 처리 → 속도 개선
# 병렬 처리 설정 (nlp.pipe())
batch_size = 100
n_process = os.cpu_count() // 2  # CPU 코어 수 조절(= 4)
dataset_name = "genia"  # DyGIE++ pretrained 모델에 맞춤
output_file = "dygie_input_bio.jsonl"

# JSONL 생성 (tqdm & 10,000개 단위로 저장)
buffer = []
save = 10000  # 10,000개 단위로 저장

with open(output_file, "a", encoding="utf8") as f:
    for idx, (doc_key, doc) in enumerate(
        tqdm(zip(doc_keys, nlp.pipe(texts, batch_size=batch_size, n_process=n_process)), total=len(texts))
    ):
        sentences = [[tok.text for tok in sent] for sent in doc.sents]
        obj = {
            "doc_key": doc_key,
            "sentences": sentences,
            "dataset": dataset_name,
            "ner": [],
            "relations": []
        }
        buffer.append(json.dumps(obj, ensure_ascii=False))

        # 10,000개 단위로 파일에 기록
        if len(buffer) >= save:
            f.write("\n".join(buffer) + "\n")
            buffer = []  # 버퍼 비우기
            f.flush()    # 디스크에 즉시 반영 (중간에 끊겨도 안전)

    # 남은 데이터도 저장
    if buffer:
        f.write("\n".join(buffer) + "\n")
        f.flush()
```

```python
# TESTING THE SCIERC MODEL OUTPUT

# test_output_scierc.jsonl 파일 확인
test_output = "C:/Users/user/Desktop/BITAmin 15/4_25-2/NLP project/test_output_scierc.jsonl"
with open(test_output, 'r', encoding="utf-8") as f:
    for line in f:
        test = json.loads(line)
        print(f"  doc_key: {test['doc_key']}")
        print(f"  dataset: {test['dataset']}")
        print(f"  sentences: {len(test['sentences'])}개")
        print(f"  ner: {test['ner']}")
        print(f"  predicted_ner: {test['predicted_ner']}, {len(test['predicted_ner'])}개")
        print(f"  relations: {test['relations']}")
        print(f"  predicted_relations: {test['predicted_relations']}, {len(test['predicted_relations'])}개")
        print(f"  predicted_clusters: {test['predicted_clusters']}\n")
```

```python
# TESTING THE SCIERC MODEL OUTPUT

# predicted_ner + predicted_relations를 기반으로 지식그래프용 삼중항 (entity1, relation, entity2) 추출

with open(test_output, "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f]
    
triples = []

for doc in data:
    sentences = doc["sentences"]
    ner_preds = doc.get("predicted_ner", [])
    rel_preds = doc.get("predicted_relations", [])

    # 전체 token flatten
    all_tokens = [token for sent in sentences for token in sent]

    # relation은 전체 token index 기준
    for rel_list in rel_preds:
        for rel in rel_list:
            # SciERC relation format: [head_start, head_end, tail_start, tail_end, rel_type, score, prob]
            head_s, head_e, tail_s, tail_e, rel_type = rel[:5]
            head_text = " ".join(all_tokens[head_s:head_e+1])
            tail_text = " ".join(all_tokens[tail_s:tail_e+1])
            triples.append((head_text, rel_type, tail_text))

# 결과 확인
for t in triples:
    print(t)
```

모델이 학습 시 6개 relation을 기준으로 문장에서 entity 쌍이 어떤 relation인지 예측하도록 학습되어 있음.

[Relation 타입 의미]
- CONJUNCTION: 두 개의 entity가 같이 묶이거나 나열됨, 연관 있음 (“and/or” 관계)
- HYPONYM-OF: 하위 개념 관계, 즉 head가 tail의 구체적 하위 개념 (ex. (“cold”,  HYPONYM-OF, “Environmental cues”) → cold는 환경적 신호의 한 종류)
- USED-FOR: head entity가 tail entity를 수행하거나 달성하는 데 사용 (ex. (“activation”, USED-FOR, “diet-induced obesity”) → activation은 diet-induced obesity를 조절하는 데 사용됨)
- PART-OF: head entity가 tail entity의 일부임, 구성 요소 관계 (ex. (“physiological and environmental stimuli”, PART-OF, “neural basis”) → 자극이 neural basis를 구성)
- FEATURE-OF: head entity가 tail entity의 특징이나 속성 (ex. (“high energy expenditure”, FEATURE-OF, “Crabp1 neurons”) → high energy expenditure는 Crabp1 neurons의 특징)
- COMPARE: 두 entity가 비교되는 관계 (ex. (“T1D beta-cells”, COMPARE, “T2D beta-cells”) → T1D와 T2D beta-cells을 서로 비교)

```python
# EXTRACTING TRIPLES FROM THE FULL OUTPUT JSONL FILE

with open(test_output, "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f]

triples = []

for doc in data:
    doc_key = doc["doc_key"]
    sentences = doc["sentences"]
    ner_preds = doc.get("predicted_ner", [])
    rel_preds = doc.get("predicted_relations", [])

    # 전체 token flatten
    all_tokens = [token for sent in sentences for token in sent]

    # relation은 전체 token index 기준
    for rel_list in rel_preds:
        for rel in rel_list:
            # SciERC relation format: [head_start, head_end, tail_start, tail_end, rel_type, score, prob]
            head_s, head_e, tail_s, tail_e, rel_type = rel[:5]
            head_text = " ".join(all_tokens[head_s:head_e+1])
            tail_text = " ".join(all_tokens[tail_s:tail_e+1])
            triples.append((doc_key, head_text, rel_type, tail_text))

# 결과 확인
for t in triples[:20]:
    print(t)

# DataFrame으로 변환 및 저장
df = pd.DataFrame(triples, columns=["doc_key", "head_entity", "relation", "tail_entity"])
df.to_csv(
    "C:/Users/user/Desktop/BITAmin 15/4_25-2/NLP project/scierc_kg_triples.csv",
    index=False,
    encoding="utf-8-sig"
)
```
